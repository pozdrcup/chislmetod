```python
import math

print("УПРАЖНЕНИЕ 1")
print("Аргументы заданы цифрами, верными в строгом смысле.")
print("Погрешность аргумента = половина единицы последнего разряда.")
print("Цифра верна в строгом смысле, если dy <= 0.5 * 10^m")
print()


def verify(name, y, dy):
    m = -10
    while 0.5 * 10 ** m < dy:
        m += 1
    if m <= 0:
        result = round(y, -m)
        digits = len(str(abs(result)).replace("0.", "").replace(".", "").lstrip("0"))
    else:
        result = round(y / 10 ** m) * 10 ** m
        digits = len(str(abs(round(y / 10 ** m))))
    print(f"{name}")
    print(f"   y  = {y:.6f}")
    print(f"   dy = {dy:.6f}")
    print(f"   верно до разряда 10^{m}  ->  {result}   ({digits} верных цифр)")
    print()


# а) lg 23,6
x, dx = 23.6, 0.05
y = math.log10(x)
dy = dx / (x * math.log(10))
verify("а) lg 23,6      f' = 1/(x*ln10)", y, dy)

# б) e^2,01
x, dx = 2.01, 0.005
y = math.exp(x)
dy = y * dx
verify("б) e^2,01       f' = e^x", y, dy)

# в) 1/4,09
x, dx = 4.09, 0.005
y = 1 / x
dy = dx / x ** 2
verify("в) 1/4,09       f' = -1/x^2", y, dy)

# г) arccos 0,79
x, dx = 0.79, 0.005
y = math.acos(x)
dy = dx / math.sqrt(1 - x ** 2)
verify("г) arccos 0,79  f' = -1/sqrt(1-x^2)", y, dy)

# д) arctg 8,45
x, dx = 8.45, 0.005
y = math.atan(x)
dy = dx / (1 + x ** 2)
verify("д) arctg 8,45   f' = 1/(1+x^2)", y, dy)

# е) 3,4^2,6 - два неточных аргумента
a, da = 3.4, 0.05
b, db = 2.6, 0.05
y = a ** b
dy_a = b * a ** (b - 1) * da
dy_b = y * math.log(a) * db
dy = dy_a + dy_b
print("е) 3,4^2,6 - здесь ДВА неточных аргумента:")
print(f"   вклад основания:  b*a^(b-1)*da = {dy_a:.6f}")
print(f"   вклад показателя: a^b*ln(a)*db = {dy_b:.6f}")
verify("   итого", y, dy)

print("ИТОГОВАЯ ТАБЛИЦА")
print(f"{'выражение':<16}{'значение':>12}{'верных цифр':>14}")
print("-" * 42)
for name, val, n in [
    ("lg 23,6", "1,37", 3),
    ("e^2,01", "7,5", 2),
    ("1/4,09", "0,244", 3),
    ("arccos 0,79", "0,7", 1),
    ("arctg 8,45", "1,453", 4),
    ("3,4^2,6", "2*10^1", 1),
]:
    print(f"{name:<16}{val:>12}{n:>14}")
```
